```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Memuat dataset (biasanya spam.csv memiliki encoding latin-1 dan kolom utama di v1 dan v2)
df = pd.read_csv('spam.csv', encoding='latin-1')

# Membersihkan dataframe jika ada kolom ekstra (seperti Unnamed: 2, dsb.)
df = df[['v1', 'v2']]
df.columns = ['label', 'text']

# Split data: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.2, random_state=42)

print("--- BAGIAN A: Multinomial Naive Bayes dengan CountVectorizer (stop_words='english') ---")
# 2. Ekstraksi fitur menggunakan CountVectorizer dengan stop_words='english'
count_vectorizer = CountVectorizer(stop_words='english')
X_train_count = count_vectorizer.fit_transform(X_train)
X_test_count = count_vectorizer.transform(X_test)

# Melatih model Multinomial Naive Bayes
nb_count = MultinomialNB()
nb_count.fit(X_train_count, y_train)

# Evaluasi model CountVectorizer
y_pred_count = nb_count.predict(X_test_count)
acc_count = accuracy_score(y_test, y_pred_count)

print(f"Akurasi CountVectorizer: {acc_count:.4f}")
print("Classification Report (CountVectorizer):")
print(classification_report(y_test, y_pred_count))


print("\n--- BAGIAN B: Multinomial Naive Bayes dengan TF-IDF (stop_words='english') ---")
# 2. Ekstraksi fitur menggunakan TF-IDF dengan stop_words='english'
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

# Melatih model Multinomial Naive Bayes
nb_tfidf = MultinomialNB()
nb_tfidf.fit(X_train_tfidf, y_train)

# Evaluasi model TF-IDF
y_pred_tfidf = nb_tfidf.predict(X_test_tfidf)
acc_tfidf = accuracy_score(y_test, y_pred_tfidf)

print(f"Akurasi TF-IDF: {acc_tfidf:.4f}")
print("Classification Report (TF-IDF):")
print(classification_report(y_test, y_pred_tfidf))


print("\n--- BAGIAN C: Kesimpulan Perbandingan ---")
print(f"Akurasi CountVectorizer: {acc_count * 100:.2f}%")
print(f"Akurasi TF-IDF          : {acc_tfidf * 100:.2f}%")
```
